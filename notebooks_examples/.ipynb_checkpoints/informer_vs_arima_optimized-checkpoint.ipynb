# Informer vs. ARIMA — busca de hiperparâmetros (`optimized`)

Este notebook é o **complemento** de `informer_vs_arima.ipynb`: mesmo protocolo de avaliação,
mesmos oito datasets e mesmas nove métricas de `notebooks_examples/calculate_metrics.ipynb`,
mas o Informer deixa de ser uma configuração fixa e passa por uma **busca aleatória de
hiperparâmetros** antes de ser comparado com o ARIMA.

O que muda em relação ao notebook original:

| | `informer_vs_arima.ipynb` | este notebook |
|---|---|---|
| Configuração da rede | fixa (`d_model=48`, `e_layers=2`, `lr=1e-3`, 6 épocas) | 40 trials aleatórios por dataset |
| Janela de decisão | — | validação cronológica interna (20% do treino) |
| Épocas do modelo final | 6 | 15 (com early stopping) |
| Baselines | ARIMA | ARIMA **e** a configuração fixa original |
| Saídas | métricas + previsões | métricas, previsões, log completo dos trials e configs vencedoras |

**Como o teste é protegido.** A série é cortada em três pedaços, sempre em ordem temporal:
treino (60%), validação (20% do treino) e teste (20%). A busca só enxerga
`(treino_interno, validação)`; o teste é usado **uma única vez**, no fim, para medir a
configuração vencedora. Nenhum trial vê o futuro do conjunto de teste.

**Custo.** A célula de busca é a mais cara do repositório: em CPU, ~20–40 s por trial nas
séries ETT. Com `SEARCH_TRIALS=40` e oito datasets, conte **algumas horas**. O log é gravado a
cada trial em `data/result/informer_tuning_trials.csv`, então a busca pode ser interrompida e
retomada (`RESUME=True`). Para uma verificação rápida, use `SEARCH_TRIALS=3` e
`DATASETS_TO_TUNE=["temperature.txt"]`.

**Dependência de código.** Este notebook usa o parâmetro `params` de
`benchmark.run_informer` (e `informer_params` de `benchmark.run_comparison`), que não existia
na versão original de `src/experiment/benchmark.py`. Sem essa mudança a célula de verificação
abaixo falha de propósito, explicando o que falta.

## 2. Três janelas: treino interno, validação e teste

`benchmark.split_series` corta treino/teste e `benchmark.split_validation` corta a cauda do
treino como validação. As duas funções preservam a ordem temporal — não há embaralhamento em
nenhum ponto do pipeline.

```python
TEST_SIZE = 0.2        # fração de teste (igual ao notebook original)
VAL_SIZE = 0.2         # fração do TREINO reservada para a busca

SEARCH_TRIALS = 40     # trials aleatórios por dataset
SEARCH_EPOCHS = 8      # épocas por trial (early stopping com patience=2)
FINAL_EPOCHS = 15      # épocas da configuração vencedora no treino completo
BASELINE_EPOCHS = 6    # épocas da configuração fixa original (comparação)
SEED = 42

# None desliga o corte por tempo. Com o orçamento, a busca sempre termina.
TIME_BUDGET_PER_DATASET = 25 * 60

RESUME = True          # retoma trials já gravados em data/result/informer_tuning_trials.csv
DATASETS_TO_TUNE = benchmark.available_datasets()

print(f"{len(DATASETS_TO_TUNE)} datasets x {SEARCH_TRIALS} trials "
      f"= {len(DATASETS_TO_TUNE) * SEARCH_TRIALS} ajustes de validação (+ ajustes finais)")
```

```python
splits = {}
rows = []
for name in DATASETS_TO_TUNE:
    spec = benchmark.DATASETS[name]
    series = benchmark.load_series(name)
    train, test = benchmark.split_series(series, TEST_SIZE)
    inner_train, validation = benchmark.split_validation(train, VAL_SIZE)
    horizon = benchmark.effective_horizon(spec, len(test))
    splits[name] = {
        "train": train,
        "test": test,
        "inner_train": inner_train,
        "validation": validation,
        "horizon": horizon,
    }
    rows.append(
        {
            "dataset": name,
            "grupo": spec.group,
            "pontos": len(series),
            "treino_interno": len(inner_train),
            "validacao": len(validation),
            "teste": len(test),
            "horizonte": horizon,
            "blocos_validacao": int(np.ceil(len(validation) / horizon)),
            "seq_len_max": len(inner_train) - horizon - 2,
            "config_original_seq_len": spec.seq_len,
        }
    )

# Garantia explícita de que a validação termina antes do teste começar.
for name, data in splits.items():
    assert data["validation"].index.max() < data["test"].index.min(), name
    assert data["inner_train"].index.max() < data["validation"].index.min(), name

df_splits = pd.DataFrame(rows).set_index("dataset")
df_splits
```

## 3. Espaço de busca

O espaço cobre os componentes que o artigo aponta como relevantes (comprimento do encoder,
largura/profundidade, distilling, fator do ProbSparse, regularização) e o que controla o
treino em CPU. `seq_len` é amostrado como **múltiplo do horizonte** para o mesmo espaço fazer
sentido nas séries horárias, de 15/30 minutos e mensais.

Restrições aplicadas na amostragem:

* `n_heads` só entra se dividir `d_model` (exigência do `MultiheadAttention` do PyTorch);
* `d_ff >= d_model`;
* `seq_len` é truncado para caber em `len(treino_interno) - horizonte - 2`;
* `label_len <= seq_len // 2` (mesma regra do `benchmark.run_informer`).

Não há `pred_len` no espaço: o decoder generativo escreve exatamente o horizonte de teste, que
é fixo por dataset — tunar isso mudaria o alvo, não o modelo.

```python
SEARCH_SPACE = {
    "seq_len_rel": [2, 3, 4, 6, 8],          # seq_len = rel * horizonte
    "label_len_rel": [0.25, 0.5],            # fração do seq_len (start token)
    "d_model": [32, 48, 64, 96],
    "n_heads": [2, 4, 8],
    "e_layers": [1, 2, 3],
    "d_layers": [1, 2],
    "d_ff": [64, 128, 192, 256],
    "dropout": [0.05, 0.1, 0.2, 0.3],
    "factor": [3, 5, 10],
    "distil": [True, False],
    "n_encoder_stacks": [1, 2],
    "lr": [1e-4, 3e-4, 5e-4, 1e-3, 3e-3],
    "weight_decay": [0.0, 1e-4, 1e-2],
    "batch_size": [32, 64],
}

print("combinações teóricas (sem restrições):",
      int(np.prod([len(v) for v in SEARCH_SPACE.values()])))
```

```python
def sample_config(rng):
    '''Amostra uma configuração válida do espaço acima (sem `key` absolutas).'''
    d_model = int(rng.choice(SEARCH_SPACE["d_model"]))
    heads = [h for h in SEARCH_SPACE["n_heads"] if d_model % h == 0] or [1]
    d_ff_choices = [ff for ff in SEARCH_SPACE["d_ff"] if ff >= d_model] or [d_model]
    return {
        "seq_len_rel": int(rng.choice(SEARCH_SPACE["seq_len_rel"])),
        "label_len_rel": float(rng.choice(SEARCH_SPACE["label_len_rel"])),
        "d_model": d_model,
        "n_heads": int(rng.choice(heads)),
        "e_layers": int(rng.choice(SEARCH_SPACE["e_layers"])),
        "d_layers": int(rng.choice(SEARCH_SPACE["d_layers"])),
        "d_ff": int(rng.choice(d_ff_choices)),
        "dropout": float(rng.choice(SEARCH_SPACE["dropout"])),
        "factor": int(rng.choice(SEARCH_SPACE["factor"])),
        "distil": bool(rng.choice(SEARCH_SPACE["distil"])),
        "n_encoder_stacks": int(rng.choice(SEARCH_SPACE["n_encoder_stacks"])),
        "lr": float(rng.choice(SEARCH_SPACE["lr"])),
        "weight_decay": float(rng.choice(SEARCH_SPACE["weight_decay"])),
        "batch_size": int(rng.choice(SEARCH_SPACE["batch_size"])),
    }


def canonical(sample):
    '''Chave estável para deduplicar trials (e sobreviver ao round-trip do CSV).'''
    return json.dumps(sample, sort_keys=True)


def build_params(sample, horizon, n_inner):
    '''Converte uma amostra em kwargs do `InformerForecaster`, ou None se não couber.'''
    seq_len = int(min(round(sample["seq_len_rel"] * horizon), n_inner - horizon - 2))
    if seq_len < horizon + 1:
        return None  # janela de treino curta demais para este horizonte
    label_len = int(max(1, min(round(sample["label_len_rel"] * seq_len), seq_len // 2)))
    params = {key: value for key, value in sample.items() if not key.endswith("_rel")}
    params.update({"seq_len": seq_len, "label_len": label_len})
    return params


# exemplo do que um trial vira na chamada do benchmark
_demo_params = build_params(sample_config(np.random.default_rng(0)), 24, 816)
_demo_params
```

## 4. Objetivo: nRMSE na validação

O objetivo é o RMSE de validação dividido pelo desvio-padrão da própria janela de validação.
Essa normalização importa porque as séries têm escalas muito diferentes (temperatura do óleo
vs. manchas solares); sem ela, um único dataset dominaria qualquer ranking agregado.

A previsão é **multi-step genuína**: `benchmark.run_informer` usa o mesmo protocolo em blocos
do notebook original (o decoder escreve o horizonte inteiro de uma vez), então o que a busca
mede é exatamente o tipo de erro que será reportado no teste.

```python
def evaluate_params(spec, inner_train, validation, horizon, params, epochs=SEARCH_EPOCHS):
    '''Treina em `inner_train` e mede o erro em `validation`. Não toca no teste.'''
    started = time.time()
    result = benchmark.run_informer(
        inner_train,
        validation,
        horizon,
        spec,
        epochs=epochs,
        verbose=False,
        params=params,
    )
    metrics = result.get("metrics") or {}
    scale = float(np.std(validation.to_numpy(dtype=float))) or 1.0
    return {
        "score_nrmse": float(metrics.get("RMSE", np.inf)) / scale,
        "val_rmse": float(metrics.get("RMSE", np.nan)),
        "val_mae": float(metrics.get("MAE", np.nan)),
        "val_smape": float(metrics.get("SMAPE", np.nan)),
        "seconds": time.time() - started,
    }
```

## 5. A busca (random search com cache)

Cada trial é gravado em `data/result/informer_tuning_trials.csv` imediatamente após terminar.
Isso dá três coisas de graça:

1. **retomada** — interromper o kernel não perde trials já feitos (`RESUME=True`);
2. **auditoria** — o log mostra toda a trajetória da busca, não só o vencedor;
3. **diagnóstico** — as células seguintes analisam sensibilidade a cada hiperparâmetro.

Trial que estoura (série curta para o `seq_len` amostrado, divisão inválida, etc.) é descartado
com aviso, não interrompe a busca.

```python
TRIALS_PATH = benchmark.RESULT_PATH / "informer_tuning_trials.csv"
BEST_CONFIGS_PATH = benchmark.RESULT_PATH / "informer_best_configs.json"

TRIAL_COLUMNS = [
    "dataset", "trial", "seq_len", "label_len", "d_model", "n_heads", "e_layers",
    "d_layers", "d_ff", "dropout", "factor", "distil", "n_encoder_stacks", "lr",
    "weight_decay", "batch_size", "score_nrmse", "val_rmse", "val_mae", "val_smape",
    "seconds", "config_json",
]
PARAM_COLUMNS = [column for column in TRIAL_COLUMNS if column not in {
    "dataset", "trial", "score_nrmse", "val_rmse", "val_mae", "val_smape", "seconds", "config_json",
}]


def load_trials():
    '''Carrega o log da busca (ou começa um vazio).'''
    if RESUME and TRIALS_PATH.exists():
        trials = pd.read_csv(TRIALS_PATH)
        print(f"retomando {len(trials)} trials de {TRIALS_PATH}")
        return trials
    return pd.DataFrame(columns=TRIAL_COLUMNS)


def save_trials(trials):
    benchmark.RESULT_PATH.mkdir(parents=True, exist_ok=True)
    trials.to_csv(TRIALS_PATH, index=False)


def tune_dataset(name, trials, n_trials=SEARCH_TRIALS, epochs=SEARCH_EPOCHS,
                 time_budget=TIME_BUDGET_PER_DATASET, verbose=True):
    '''Random search determinístico para um dataset; devolve o log atualizado.'''
    spec = benchmark.DATASETS[name]
    data = splits[name]
    inner_train, validation, horizon = data["inner_train"], data["validation"], data["horizon"]

    rng = np.random.default_rng(SEED + zlib.crc32(name.encode()))
    done = set(trials.loc[trials["dataset"] == name, "config_json"])
    placed = int((trials["dataset"] == name).sum())
    attempts = 0
    deadline = None if time_budget is None else time.time() + time_budget

    while placed < n_trials:
        attempts += 1
        if deadline is not None and time.time() > deadline:
            print(f"  [{name}] orçamento de {time_budget / 60:.0f} min esgotado "
                  f"({placed}/{n_trials} trials)")
            break
        if attempts > 50 * n_trials:
            print(f"  [{name}] muitas amostras inválidas; parando em {placed}/{n_trials}")
            break

        sample = sample_config(rng)
        key = canonical(sample)
        if key in done:
            continue
        done.add(key)

        params = build_params(sample, horizon, len(inner_train))
        if params is None:
            continue
        try:
            outcome = evaluate_params(spec, inner_train, validation, horizon, params, epochs=epochs)
        except Exception as error:
            print(f"  [{name}] trial descartado ({type(error).__name__}: {error})")
            continue

        row = {"dataset": name, "trial": placed}
        row.update({column: params[column] for column in PARAM_COLUMNS})
        row.update(outcome)
        row["config_json"] = key
        trials = pd.concat([trials, pd.DataFrame([row])], ignore_index=True)
        placed += 1

        if verbose:
            print(f"  [{name}] {placed:3d}/{n_trials} nRMSE={outcome['score_nrmse']:.4f} "
                  f"({outcome['seconds']:.1f}s) size={params['seq_len']}/{params['d_model']}x"
                  f"{params['e_layers']} lr={params['lr']:g}")
        save_trials(trials)

    return trials
```

## 6. Executando a busca

Para cada dataset: 40 trials aleatórios, com corte de 25 min por dataset. Reduza
`SEARCH_TRIALS`/`TIME_BUDGET_PER_DATASET` na célula de configuração se quiser uma passada
rápida. O `print` mostra o nRMSE de validação de cada trial conforme ele termina.

```python
all_trials = load_trials()

for name in DATASETS_TO_TUNE:
    print(f"== {name} (validação com {len(splits[name]['validation'])} pontos, "
          f"horizonte {splits[name]['horizon']})")
    all_trials = tune_dataset(name, all_trials)

print(f"\ntotal de trials no log: {len(all_trials)}")
all_trials.groupby("dataset")["score_nrmse"].agg(["count", "min"]).round(4)
```

### 6.1 Os melhores trials de cada dataset

```python
df_trials = all_trials.dropna(subset=["score_nrmse"]).copy()

display_columns = [
    "dataset", "trial", "seq_len", "label_len", "d_model", "n_heads", "e_layers",
    "d_layers", "d_ff", "dropout", "factor", "distil", "lr", "weight_decay",
    "batch_size", "score_nrmse",
]
df_top = (
    df_trials.sort_values("score_nrmse")
    .groupby("dataset", sort=False)
    .head(3)[display_columns]
    .reset_index(drop=True)
)
df_top.round(4)
```

### 6.2 Sensibilidade: como cada hiperparâmetro move o erro

Média do nRMSE de validação por valor de hiperparâmetro, no dataset mais longo da busca.
Ajuda a separar o que realmente importou (ex.: `lr`, `seq_len`) do que foi ruído de uma única
janela de validação — e é o tipo de leitura que um único resultado de teste não permite.

```python
diagnostic_dataset = DATASETS_TO_TUNE[0]
subset = df_trials[df_trials["dataset"] == diagnostic_dataset]
print(f"dataset: {diagnostic_dataset} | trials: {len(subset)}")

for column in ["seq_len", "d_model", "e_layers", "dropout", "lr", "batch_size", "distil"]:
    table = (
        subset.groupby(column)["score_nrmse"]
        .agg(["mean", "min", "count"])
        .sort_values("mean")
        .round(4)
    )
    print(f"\n-- {column}")
    print(table.to_string())
```

### 6.3 Configuração vencedora por dataset

Escolhe o trial de menor nRMSE de validação e reconstrói o dicionário de parâmetros a partir do
`config_json` (assim booleans e floats voltam com o tipo certo, mesmo depois de um round-trip
pelo CSV). O arquivo `data/result/informer_best_configs.json` guarda o resultado.

```python
best_params = {}
best_rows = []

for name in DATASETS_TO_TUNE:
    subset = df_trials[df_trials["dataset"] == name]
    if subset.empty:
        print(f"  [{name}] sem trials válidos; mantendo a configuração original")
        continue
    row = subset.loc[subset["score_nrmse"].idxmin()]
    sample = json.loads(row["config_json"])
    params = build_params(sample, splits[name]["horizon"], len(splits[name]["inner_train"]))
    best_params[name] = params
    best_rows.append({"dataset": name, **params, "val_nrmse": row["score_nrmse"],
                      "val_rmse": row["val_rmse"], "trial": int(row["trial"])})

benchmark.RESULT_PATH.mkdir(parents=True, exist_ok=True)
BEST_CONFIGS_PATH.write_text(json.dumps(best_params, indent=2, default=str))
print("configurações salvas em:", BEST_CONFIGS_PATH)

df_best = pd.DataFrame(best_rows).set_index("dataset")
df_best.round(4)
```

## 7. Avaliação final no teste

Agora o teste é usado pela primeira vez. Três modelos, todos no mesmo protocolo em blocos de
`horizonte` passos:

* **`arima`** — `auto_arima` do projeto (`benchmark.run_arima`), inalterado;
* **`informer_baseline`** — a configuração fixa do notebook original (`params=None`,
  6 épocas), para isolar o efeito da busca;
* **`informer_opt`** — a configuração vencedora, retreinada no treino completo (não no treino
  interno) com `FINAL_EPOCHS=15` épocas e early stopping.

A célula abaixo usa `benchmark.run_comparison(informer_params=...)`: ARIMA e Informer tunado
rodam juntos, com as métricas no mesmo layout de `metrics.open_fold_result`.

```python
df_opt, results_opt = benchmark.run_comparison(
    dataset_names=DATASETS_TO_TUNE,
    test_size=TEST_SIZE,
    epochs=FINAL_EPOCHS,
    informer_params=best_params,
    verbose=True,
)
df_opt.round(4)
```

### 7.1 Baseline fixo (o Informer do notebook original)

Só o Informer é reajustado aqui — o resultado do ARIMA já está em `df_opt`, então não há
motivo para estimar o `auto_arima` duas vezes.

```python
rows = []
baseline_results = {}

for name in DATASETS_TO_TUNE:
    spec = benchmark.DATASETS[name]
    train, test = splits[name]["train"], splits[name]["test"]
    horizon = splits[name]["horizon"]
    result = benchmark.run_informer(
        train, test, horizon, spec, epochs=BASELINE_EPOCHS, verbose=False
    )
    baseline_results[name] = result
    rows.append(
        {
            "ts": name,
            "model": "informer_baseline",
            **{column: result["metrics"].get(column, np.nan) for column in benchmark.METRIC_COLUMNS},
            "time_training": result["time_training"],
            "time_testing": result["time_testing"],
        }
    )

df_baseline = pd.DataFrame(rows).set_index(["ts", "model"])
df_baseline = df_baseline[benchmark.METRIC_COLUMNS + ["time_training", "time_testing"]]

df_tuned = df_opt.reset_index().replace({"model": {"informer": "informer_opt"}})
df_final = (
    pd.concat([df_tuned, df_baseline.reset_index()])
    .set_index(["ts", "model"])[benchmark.METRIC_COLUMNS + ["time_training", "time_testing"]]
)
df_final.round(4)
```

## 8. Resultados

`ganho_tuning_%` positivo = a busca reduziu o erro em relação ao Informer fixo;
`ganho_vs_arima_%` positivo = o Informer tunado ficou à frente do ARIMA naquela série.
Preste atenção nos dois separadamente: eles respondem perguntas diferentes ("a busca ajudou?" e
"o modelo ganha do baseline?").

```python
for metric in ["RMSE", "MAE", "SMAPE"]:
    pivot = df_final[metric].unstack("model")
    table = pd.DataFrame({
        "arima": pivot["arima"],
        "informer_baseline": pivot["informer_baseline"],
        "informer_opt": pivot["informer_opt"],
        "melhor": pivot.idxmin(axis=1),
        "ganho_tuning_%": (pivot["informer_baseline"] - pivot["informer_opt"])
        / pivot["informer_baseline"] * 100,
        "ganho_vs_arima_%": (pivot["arima"] - pivot["informer_opt"]) / pivot["arima"] * 100,
    })
    print("=" * 90)
    print(metric)
    print(table.round(3).to_string())

df_gain = pd.DataFrame({
    metric: (df_final[metric].unstack("model")["informer_baseline"]
             - df_final[metric].unstack("model")["informer_opt"])
    / df_final[metric].unstack("model")["informer_baseline"] * 100
    for metric in ["RMSE", "MAE"]
})
df_gain.round(3)
```

### 8.1 Erro normalizado (comparável entre séries de escalas diferentes)

```python
rows = []
for name, data in results_opt.items():
    if "test" not in data:
        continue
    scale = float(np.std(data["test"].to_numpy(dtype=float))) or 1.0
    rows.append(
        {
            "ts": name,
            "arima": data["arima"]["metrics"]["RMSE"] / scale,
            "informer_opt": data["informer"]["metrics"]["RMSE"] / scale,
            "informer_baseline": baseline_results[name]["metrics"]["RMSE"] / scale,
        }
    )

df_nrmse = pd.DataFrame(rows).set_index("ts")
df_nrmse.round(4)
```

## 11. Como ler os resultados (e o que a busca não resolve)

**O que a busca deve mostrar.** Se a hipótese do notebook original estava certa, o ganho do
tuning é maior nas séries longas e sazonais (ETT, `taylor`): mais dados, mais espaço para
`seq_len` grande, `d_model` maior e `lr` menor pagarem. Nas séries curtas
(`sunspot`, `temperature`, `coloradoRiver`) o ganho tende a ser pequeno ou nulo — e pode até
ser negativo no teste, porque a validação tem só 2–4 blocos.

**Assimetria de orçamento — leia com cuidado.** O `informer_opt` tem 15 épocas contra 6 do
`informer_baseline`. Parte do ganho vem dos hiperparâmetros, parte vem de treinar mais. A coluna
`ganho_tuning_%` mistura os dois efeitos; ela responde "o pipeline tunado é melhor?", não
"qual hiperparâmetro causou isso". Para separar, rode o baseline com `BASELINE_EPOCHS=15`.

**Limitações da busca:**

1. **uma única janela de validação** (20% final do treino) por dataset — 2 a 6 blocos de
   previsão. O ranking de 40 trials sobre uma amostra tão pequena tem ruído considerável; o
   top-1 pode não ser o melhor modelo de verdade;
2. **sem validação cruzada aninhada**: o número de trials não foi escolhido por CV, e não há
   segunda janela de validação independente;
3. **semente única** por trial (`seed=42` herdado do wrapper). Variância de inicialização não é
   medida; repetir os 3 melhores trials com sementes diferentes seria o próximo passo natural;
4. **corte por tempo** (`TIME_BUDGET_PER_DATASET`) favorece trials rápidos: redes grandes podem
   ter sido avaliadas em menos épocas efetivas, o que enviesa o ranking contra elas;
5. **o teste continua sendo uma única partição**: comparar com o ARIMA numa só janela de 20% não
   autoriza conclusão sobre superioridade geral do modelo;
6. **MAPE** pode ser `inf` (alvo com zeros) — use RMSE, MAE, SMAPE, IA;
7. a rede segue **muito menor** que a do artigo (`d_model=512`, 8 épocas com grid search,
   GPU). Este notebook melhora o ajuste dentro de um orçamento de CPU, não reproduz o setup
   do paper.

**Próximos passos, em ordem de retorno:**

1. repetir os top-3 trials com 3 sementes e escolher pela média (barato, ataca a limitação 3);
2. segunda janela de validação (walk-forward com 2 folds) para medir estabilidade do ranking;
3. só então aumentar o espaço de busca (ex.: `seq_len` até 12× o horizonte).
