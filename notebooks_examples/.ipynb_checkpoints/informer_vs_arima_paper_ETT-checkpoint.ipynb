# Informer no ETT — protocolo do artigo (Tabela 2)

Este notebook reproduz o experimento **multivariado de ETT** de Zhou et al. (2021),
*Informer: Beyond Efficient Transformer for Long Sequence Time-Series Forecasting*
(`docs/2012.07436v3.pdf`), e compara o resultado com o ARIMA do projeto
(`src/model/arima.py`).

O que muda em relação a `informer_vs_arima.ipynb`:

* as **7 variáveis** do ETT (`HUFL, HULL, MUFL, MULL, LUFL, LULL, OT`) são entrada **e**
  saída (7 → 7), como na Tabela 2 do artigo, em vez de prever apenas `OT` com janelas curtas;
* split **12/4/4 meses**, horizontes de **24 a 720** pontos e avaliação *rolling* com
  **stride 1** (uma janela por origem do teste), como na seção 4 do artigo;
* rede e treino do artigo: `d_model=512`, `d_ff=2048`, encoder 3+1 camadas, decoder
  2 camadas, Adam `1e-4` com decaimento de 0.5 por época, 8 épocas, batch 32.

Os detalhes do protocolo estão na seção 1 e as decisões de reprodução (e seus desvios)
na seção 8. O experimento é **caro**: são 20 combinações (4 datasets x 5 horizontes) e,
no protocolo fiel, cada uma treina uma rede de ~22,8 milhões de parâmetros.

> **Antes de executar:** rode as células em ordem. A célula de execução está protegida por
> `RUN_EXPERIMENT = False`; mude para `True` quando quiser rodar (veja a seção 5).
> Os resultados ficam em `data/result/paper_ett/` e a execução é **retomável**.

## 1. Protocolo do artigo

| Item | Valor usado | Onde está no artigo |
| --- | --- | --- |
| Entrada / saída | 7 variáveis → 7 variáveis (M) | Tabela 2 (multivariado); seção 4 ("Multivariate") |
| Split | 12 meses treino / 4 validação / 4 teste | seção 4, "Datasets" |
| Horizontes ETTh | 24, 48, 168, 336, 720 | Tabela 1/2 e seção 4 ("{1d, 2d, 7d, 14d, 30d}") |
| Horizontes ETTm | 24, 48, 96, 288, 672 | Tabela 1/2 e seção 4 ("{6h, 12h, 24h, 72h, 168h}") |
| Avaliação | janela de `h` passos por origem, **stride 1** | seção 4, "Metrics" ("roll the whole set with stride = 1") |
| Métricas | MSE e MAE por janela, média sobre janelas e variáveis | seção 4, "Metrics" |
| Encoder | stack principal de 3 camadas + réplica de 1 camada com 1/4 da entrada | seção 4.2 e Tabela 7 |
| Decoder | 2 camadas, geração de todo o horizonte em uma passada | seção 4.2 e Tabela 7 |
| Dimensões | `d_model=512`, `d_ff=2048`, dropout 0.1, fator `c=5` | Tabela 7 |
| Treino | MSE, Adam `1e-4` (metade por época), 8 épocas, batch 32, early stopping | seção 4.2 e Apêndice E |
| Normalização | cada variável com média 0 e desvio 1 (estatísticas do treino) | seção 4.2 ("zero-mean normalized") |

O ARIMA **não aparece na Tabela 2** do artigo; ele entra aqui como baseline extra do projeto,
estimado por variável (7 modelos univariados) e avaliado nas mesmas janelas do Informer.

## 5. Configuração da execução

O escopo padrão é o do artigo (todos os datasets e horizontes). Para um teste rápido,
reduza `DATASETS_TO_RUN`, `HORIZONS_TO_RUN`, use `CONFIG_OVERRIDES` com uma rede menor e
`MAX_ORIGINS` para limitar as janelas de avaliação.

Custos aproximados (GPU; em CPU pode ser 10-20x mais lento):

* treino do Informer por (dataset, horizonte): 8 épocas sobre 8 mil janelas (ETTh) ou
  34 mil (ETTm); em GPU, minutos por combinação;
* avaliação stride 1: ~2.880 janelas no ETTh e ~11.500 no ETTm, processadas em lotes
  (rápido, poucos minutos por combinação);
* ARIMA: 7 `auto_arima` por dataset (minutos; `m=24` no ETTh é mais lento) e a avaliação
  rolling com `predict(h)` a cada origem, que é a parte mais cara.

Se faltar memória nos horizontes longos (720/672), reduza `batch_size` (32 → 16 ou 8) em
`CONFIG_OVERRIDES`: o decoder generativo processa `label_len + h` posições por amostra.

`RUN_EXPERIMENT` começa em `False` de propósito. As combinações já concluídas são lidas de
`data/result/paper_ett/metrics.csv` e os pesos de
`data/result/paper_ett/checkpoints/`, então interromper e reexecutar não repete trabalho.

```python
# --- escopo do artigo -------------------------------------------------------
DATASETS_TO_RUN = paper.available_datasets()
HORIZONS_TO_RUN = None          # None -> todos os horizontes do dataset
MODELS = ("informer", "arima")

# --- execução ---------------------------------------------------------------
RUN_EXPERIMENT = False          # mude para True para rodar
FORCE_RERUN = False             # True -> ignora métricas e checkpoints em cache
                                # (uma config customizada já ignora o cache de métricas)
MAX_ORIGINS = None              # inteiro pequeno -> smoke test (não é o protocolo do artigo)
CONFIG_OVERRIDES = None         # ex.: {"epochs": 1, "d_model": 64, "n_heads": 4,
                                #      "d_ff": 128, "e_layers": 2, "d_layers": 1,
                                #      "n_encoder_stacks": 2, "replica_layers": 1}
ARIMA_SEASONAL_PERIOD = None    # None -> sem sazonalidade; 24 -> diária no ETTh (mais lento)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", DEVICE)
print("datasets:", DATASETS_TO_RUN)
print("modelos:", MODELS)
print("executar:", RUN_EXPERIMENT)
```

## 6. Execução

Cada combinação treina (ou carrega do checkpoint) o Informer, avalia o teste inteiro com
stride 1, ajusta/carrega os 7 ARIMA e avalia as mesmas janelas. As linhas são gravadas
incrementalmente em `data/result/paper_ett/metrics.csv`, então a célula é retomável.

O ARIMA é avaliado no mesmo recorte de teste, mas com um modelo univariado por variável
(os 7 modelos do projeto), como baseline extra.

```python
if RUN_EXPERIMENT:
    rows = []
    for name in DATASETS_TO_RUN:
        horizons = HORIZONS_TO_RUN or paper.DATASETS[name].horizons
        for horizon in horizons:
            print(f"\n=== {name} | h={horizon} ===")
            rows.extend(
                paper.run_experiment(
                    name,
                    horizon,
                    models=MODELS,
                    config=CONFIG_OVERRIDES,
                    device=DEVICE,
                    force=FORCE_RERUN,
                    max_origins=MAX_ORIGINS,
                    arima_seasonal_period=ARIMA_SEASONAL_PERIOD,
                    verbose=True,
                )
            )
    run_frame = pd.DataFrame(rows)
    display(run_frame[["dataset", "horizon", "model", "MSE", "MAE",
                       "MSE_OT", "MAE_OT", "n_windows", "time_training", "time_testing"]].round(4))
else:
    print("RUN_EXPERIMENT = False — nada foi executado.")
    print("Defina RUN_EXPERIMENT = True na célula de configuração para rodar o protocolo.")
```

## 7. Resultados

`MSE`/`MAE` seguem o protocolo do artigo (média sobre as 7 variáveis e sobre as janelas).
`MSE_OT`/`MAE_OT` isolam a variável alvo do projeto e ajudam a comparar com o ARIMA.
`MSE_paper`/`MAE_paper` são os números da Tabela 2 (apenas como referência: treino em GPU,
5 execuções e código oficial).

```python
metrics = paper.load_metrics()
if metrics.empty:
    print("Sem resultados em", paper.metrics_path())
else:
    table = paper.comparison_table(metrics)
    table = table.merge(paper.reference_table(), how="left", on=["dataset", "horizon"])
    if {"MSE_arima", "MSE_informer"}.issubset(table.columns):
        table["informer_gain_%"] = (table["MSE_arima"] - table["MSE_informer"]) / table["MSE_arima"] * 100
        table["informer_vs_paper_%"] = (table["MSE_informer"] - table["MSE_paper"]) / table["MSE_paper"] * 100
    display(table.sort_values(["dataset", "horizon"]).round(4))
```

```python
if not metrics.empty:
    for name in metrics["dataset"].unique():
        subset = metrics[metrics["dataset"] == name]
        figure, axis = plt.subplots(figsize=(8, 3.4))
        for model in subset["model"].unique():
            part = subset[subset["model"] == model].sort_values("horizon")
            axis.plot(part["horizon"], part["MSE"], marker="o", label=model.upper())
        reference = paper.PAPER_REFERENCE.get(name)
        if reference:
            horizons = sorted(reference)
            axis.plot(horizons, [reference[h][0] for h in horizons], "k--", marker="x",
                      label="Informer (Tabela 2)")
        axis.set_title(f"{name} — MSE médio (7 variáveis, stride 1)")
        axis.set_xlabel("horizonte")
        axis.set_ylabel("MSE")
        axis.set_xscale("log")
        axis.set_xticks(sorted(subset["horizon"].unique()))
        axis.set_xticklabels(sorted(subset["horizon"].unique()))
        axis.legend(fontsize=8)
        figure.tight_layout()
        plt.show()
```

### Janelas de exemplo

Para cada dataset/horizonte executado, o primeiro e o último recorte do teste foram salvos
em `data/result/paper_ett/samples/` (só a variável `OT` é plotada).

```python
if not metrics.empty:
    for name in metrics["dataset"].unique():
        horizons = sorted(metrics.loc[metrics["dataset"] == name, "horizon"].unique())
        horizon = int(horizons[-1])
        informer_samples = paper.load_samples(name, horizon, "informer")
        if informer_samples is None:
            continue
        figure, axis = plt.subplots(figsize=(10, 3))
        axis.plot(np.arange(horizon), informer_samples["true"][0][:, 6],
                  color="black", linewidth=1.2, label="real")
        for model, color in (("informer", "tab:blue"), ("arima", "tab:orange")):
            samples = paper.load_samples(name, horizon, model)
            if samples is None:
                continue
            axis.plot(np.arange(horizon), samples["pred"][0][:, 6], color=color,
                      linewidth=1.0, label=model.upper())
        axis.set_title(f"{name} — primeira janela do teste (h={horizon}, OT)")
        axis.legend(fontsize=8)
        figure.tight_layout()
        plt.show()
```

## 8. Como interpretar e o que difere do artigo

**O que deve bater com a Tabela 2:** a ordem de grandeza do MSE/MAE em cada horizonte, o
crescimento do erro com `h` e a média sobre as 7 variáveis. Números absolutos dificilmente
serão idênticos: o artigo reporta média de **5 execuções** com reamostragem de treino/validação
e treinou em V100, enquanto aqui há **uma execução** com split fixo.

**Desvios conscientes desta reprodução:**

1. **Atenção com 8 cabeças** em todo o modelo. A Tabela 7 descreve `h=16, d=32` para o
   ProbSparse e `h=8, d=64` na atenção cruzada; o `Informer` do projeto usa um único
   `n_heads` para as duas, então foi usado 8 (mesma escolha do repositório oficial).
2. **Uma execução, split fixo** e semente 42. O artigo faz 5 execuções com *shifts* de
   validação; implementar isso aqui multiplicaria o custo por 5.
3. **ETTm2** não aparece nas tabelas do artigo (ele usa ETTh1, ETTh2 e ETTm1), mas foi
   incluído porque o dataset acompanha o ETT e usa o mesmo protocolo.
4. **ARIMA** é o `pmdarima.auto_arima` do projeto, um modelo por variável. A Tabela 2 não
   reporta ARIMA (apenas LSTnet e as variantes do Informer), então essa comparação é um
   extra do projeto. Por padrão ele é **não sazonal** (rápido); use
   `ARIMA_SEASONAL_PERIOD = 24` para sazonalidade diária no ETTh, ciente de que o
   `auto_arima` fica bem mais lento.
5. **Dropout 0.1** (Tabela 7); o repositório oficial usa 0.05 nos scripts de ETT.
6. A amostragem do ProbSparse é aleatória a cada forward; a semente é fixada antes da
   avaliação para que as janelas sejam reprodutíveis.

**Limitações:** `MSE`/`MAE` não são comparáveis com as métricas de 1 passo do notebook
`informer_vs_arima.ipynb`; o protocolo aqui é multi-step com stride 1. O ARIMA usa apenas os
meses de treino (sem validação) e não tem a mesma capacidade dos modelos do artigo.

Para limpar um recorte e refazer: `paper.clear_cache(name="ETTh2.csv", horizon=24)`.
Para limpar tudo: apague `data/result/paper_ett/` (métricas, checkpoints, modelos ARIMA).
